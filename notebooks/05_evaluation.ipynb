Recall@k on the hand-labeled `data/eval/qa_pairs.json` set, plus manual faithfulness spot-checks.

```python
import sys, os, json
sys.path.append('..')
from dotenv import load_dotenv
load_dotenv('../.env')

from src.indexing import load_indexes
from src.evaluate import recall_at_k, summarize

text_index, image_index, text_metadata, image_metadata = load_indexes('../indexes')

with open('../data/eval/qa_pairs.json') as f:
    qa_pairs = json.load(f)
print(len(qa_pairs), 'eval questions loaded')
```
